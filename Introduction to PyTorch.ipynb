# 4. Linear Regression Scratch

Create the data

```python
n = 10
d_in = 1
d_out = 1

x = torch.randn(n, d_in)

true_w = torch.tensor([[2.0]])
true_b = torch.tensor([1.0])

y_true = x @ true_w + true_b

print(f"n = {n}, d_in = {d_in}, d_out = {d_out}")
print(f"x.shape      = {tuple(x.shape)}")
print(f"y_true.shape = {tuple(y_true.shape)}")
print(f"true_w = {true_w.item()}, true_b = {true_b.item()}   (the secret answers)")
```

Initializing parameters

```python
w = torch.randn(d_in, d_out, requires_grad=True)
b = torch.randn(d_out, requires_grad=True)

print(f"initial w = {w.item():.4f}   (random guess)")
print(f"initial b = {b.item():.4f}   (random guess)")
print(f"w.requires_grad = {w.requires_grad}")
print(f"b.requires_grad = {b.requires_grad}")
```

The forward pass

```python
y_hat = x @ w + b

print("y_hat (predictions):")
print(y_hat)
print("\ny_true (the truth):")
print(y_true)

print("\nNotice y_hat has a grad_fn -- autograd is watching.")
print(f"y_hat.grad_fn = {y_hat.grad_fn}")
```

The Loss

```python
error = y_hat - y_true
squared_error_loss = error ** 2
loss = squared_error_loss.mean()

print(f"loss = {loss.item():.4f}")
print(f"loss.grad_fn = {loss.grad_fn}")
print("The loss sits at the very end of the computation graph.")
print("It is the source from which all knowledge will flow backward.")
```

The Backward Pass

```python
loss.backward()

print(f"w.grad = {w.grad.item():.4f}")
print(f"b.grad = {b.grad.item():.4f}")
print()
print("The sign is everything:")
print("  A NEGATIVE gradient means increasing the parameter would DECREASE the loss.")
print("  The gradient always points toward the steepest INCREASE.")
print("  So to minimize the loss, we go in the OPPOSITE direction.")
```

Manual Gradient Descent

```python
learning_rate = 0.01

with torch.no_grad(): # Makes autograd not track this step
  w -= learning_rate * w.grad
  b -= learning_rate * b.grad

print(f"w updated: {w.item():.4f}")
print(f"b updated: {b.item():.4f}")

w.grad.zero_()
b.grad.zero_()

print(f"w.grad after zeroing: {w.grad.item():.4f}")
print(f"b.grad after zeroing: {b.grad.item():.4f}")
```

The full training loop

```python
torch.manual_seed(42)

w = torch.randn(d_in, d_out, requires_grad=True)
b = torch.randn(d_out, requires_grad=True)

epochs = 100
learning_rate = 0.01

for epoch in range(epochs):
  y_hat = x @ w + b

  error = y_hat - y_true
  squared_error_loss = error ** 2
  loss = squared_error_loss.mean()

  loss.backward()

  with torch.no_grad():
    w -= learning_rate * w.grad
    b -= learning_rate * b.grad

  w.grad.zero_()
  b.grad.zero_()

  if epoch % 10 == 0:
        print(f"{epoch:>6} | {loss.item():>10.4f} | {w.item():>10.4f} | {b.item():>10.4f}")
```

```python
print(f"learned w = {w.item():.4f}   (true w = {true_w.item():.4f})")
print(f"learned b = {b.item():.4f}   (true b = {true_b.item():.4f})")
print()
print("We built a learning machine from scratch.")
```

# 6. Advanced Layers

nn.Embedding: Turn words into numbers

```python
vocab_size = 10
embedding_dim = 3

embedding_layer = nn.Embedding(vocab_size, embedding_dim)
input_ids = torch.tensor([[1, 5, 9, 8]])

word_vectors = embedding_layer(input_ids)

print(f"vocab_size = {vocab_size}, embedding_dim = {embedding_dim}")
print(f"input_ids.shape    = {tuple(input_ids.shape)}")
print(f"word_vectors.shape = {tuple(word_vectors.shape)}")
print()
print("For each of our four input ids, it looked up and returned the")
print("corresponding three-dimensional vector.")
print()
print("word_vectors:")
print(word_vectors)
print()
print("This is the first step in every large language model.")
```

nn.LayerNorm: The traffic cap

```python
norm_layer = nn.LayerNorm(normalized_shape=3) # 3 is just the number of columns

sample_features = torch.randn(2, 3)
normalized = norm_layer(sample_features)

print("sample_features:")
print(sample_features)
print(f"\nmean per row: {sample_features.mean(dim=-1).tolist()}")
print(f"std  per row: {sample_features.std(dim=-1).tolist()}")
print()
print("normalized:")
print(normalized)
print(f"\nmean per row: {normalized.mean(dim=-1).tolist()}   <- basically zero")
print(f"std  per row: {normalized.std(dim=-1, unbiased=False).tolist()}   <- basically one")
print()
print("It keeps the network stable. It's beautiful.")
```

nn.Dropout: Prevent Overfitting

```python
dropout_layer = nn.Dropout(p=0.5)
input_tensor = torch.ones(5)

print("input_tensor (all ones):")
print(input_tensor)

dropout_layer.train()
train_output = dropout_layer(input_tensor)
print("\n--- dropout_layer.train() ---")
print("train_output:")
print(train_output)
print("Randomly zeroed and scaled. Some values are zero, others are two.")

dropout_layer.eval()
eval_output = dropout_layer(input_tensor)

print("\n--- dropout_layer.eval() ---")
print("eval_output:")
print(eval_output)
print("It's the identity function. It just passes the original tensor through.")
print()
print("This simple on/off switch is a cornerstone of modern deep learning.")
```

#7. The Professional Result

```python
class LinearRegressionModel(nn.Module):
  def __init__(self, in_features, out_features):
    super().__init__()
    self.linear_layer = nn.Linear(in_features, out_features)

  def forward(self, x):
    y_hat = self.linear_layer(x)
    return y_hat

model = LinearRegressionModel(in_features = 1, out_features = 1)
print("Model Architecture : ")
print(model)
```

The Optimizer

```python
learning_rate = 0.01
optimizer = optim.Adam(model.parameters(), lr = learning_rate)
loss_fn = nn.MSELoss()
```

```python
epochs = 100
x = torch.randn(10, 1)

true_w = torch.tensor([[2.0]])
true_b = torch.tensor([1.0])

y_true = x @ true_w + true_b

for epoch in range(epochs):
  y_hat = model(x)
  loss = loss_fn(y_hat, y_true)

  optimizer.zero_grad()
  loss.backward()
  optimizer.step()

  if epoch % 10 == 0:
    print(f"Epoch {epoch:02d}: Loss = {loss.item():.4f}")
```

## The whole pipeline in basically 5 steps:

### 1. Compute y-hat.
### 2. Find error.
### 3. optimizer.zero_grad()
### 4. loss.backward()
### 5. optimizer.step()
